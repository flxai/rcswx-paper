```python
import einsearch
import torch
from torch.utils.data import DataLoader
from torchvision.datasets import CIFAR10
from torchvision import transforms
from PIL import Image

# Define your transformation pipeline
transform = transforms.Compose([
    transforms.PILToTensor(),  # Convert PIL Image to tensor (uint8, shape: CxHxW)
    transforms.ConvertImageDtype(torch.float32)  # Convert to float32
])

train_loader = DataLoader(CIFAR10(root = "cifar10", train = True, transform = transform), shuffle=True, batch_size=16)
val_loader = DataLoader(CIFAR10(root = "cifar10", train = False, transform = transform), shuffle=False, batch_size=16)
```

```python
einsearch.search(train_loader = val_loader, val_loader = val_loader, verbose_setup = True, device = "cpu", steps = 1, population_size = 2, epochs = 2, test_epochs = 2)
```
